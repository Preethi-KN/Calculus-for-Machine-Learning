```python
def f(x):
    return x**3 - 2*x**2 + x   # f(x) = x³ - 2x² + x

def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)   # central difference

x_vals = [0, 1, 2, 3]
print(f"   f(x) = x³ - 2x² + x")
print(f"   f'(x) analytical = 3x² - 4x + 1")
print(f"\n   {'x':>4}  {'f(x)':>8}  {'f′(x) numerical':>18} ")
print("   " + "-" * 50)
for x in x_vals:
    exact = 3*x**2 - 4*x + 1
    numerical = numerical_derivative(f, x)
    print(f"   {x:>4}  {f(x):>8.4f}  {numerical:>18.6f} ")
```

```python
# ── CHAIN RULE ─────────────────────────────────────────────
print("\n── 5. CHAIN RULE — Backprop Example ──────────────────")
print("   h(x) = sin(x²)  →  h′(x) = cos(x²)·2x")
x_val = np.pi / 4
h_prime = np.cos(x_val**2) * 2 * x_val
print(f"   At x = π/4:  h′(x) = {h_prime:.6f}")

# ── LINEAR REGRESSION WITH GRADIENT DESCENT ───────────────
print("\n── 6. LINEAR REGRESSION — Gradient Descent ───────────")
np.random.seed(42)
X   = np.array([1, 2, 3, 4, 5], dtype=float)
y   = np.array([2, 4, 5, 4, 6], dtype=float)
w, b = 0.0, 0.0
alpha = 0.01
n = len(X)

for _ in range(1000):
    y_pred = w * X + b
    dw = (-2/n) * np.sum(X * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    w -= alpha * dw
    b -= alpha * db

print(f"   Data: X = {X.tolist()},  y = {y.tolist()}")
print(f"   After 1000 iterations:")
print(f"   w = {w:.4f},  b = {b:.4f}")
print(f"   Predictions: {(w * X + b).round(4).tolist()}")
mse = np.mean((y - (w * X + b))**2)
print(f"   Final MSE   = {mse:.6f}")
```
